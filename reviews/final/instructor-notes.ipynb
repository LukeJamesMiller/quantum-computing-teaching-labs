# Common Pitfalls in QML Design
## Circuit Design Errors:
- Incorrect feature encoding (e.g., wrong angles in ZZ map)
- Overly deep ansatz leading to noise accumulation


## Optimization Issues:
- Poor parameter initialization causing barren plateaus
- Ignoring noise in expectation value computations


## Data Handling:
- Inefficient encoding for high-dimensional data
- Misinterpreting quantum measurements (e.g., shot noise)


## Application Missteps:
- Assuming quantum advantage without validation
- Neglecting classical benchmarks in QML


## Mitigation:
Use shallow circuits, error mitigation, hybrid workflows

## Solution:

#  Practice Problem 2: QSVM

## Problem:
Derive quantum kernel for 2-qubit ZZ feature map
## Solution:
- Feature map: ( $U_\phi(x) = \prod_i e^{-i x_i Z_i} \prod_{i<j} e^{-i (x_i - x_j)^2 Z_i Z_j}$ )
- Kernel: ( $K(x_i, x_j) = |\langle 0 | U_\phi^\dagger(x_i) U_\phi(x_j) | 0 \rangle|^2$ )
- Circuit: Apply ( $U_\phi(x_j) ), then ( U_\phi^\dagger(x_i)$ ), measure


##  Task:
Explain why entanglement aids separability
## AI Relevance:
Enhanced classification in complex datasets

# Practice Problem 3: Quantum Feature Selection

## Problem:
Outline a quantum MI-based feature selection for 3 features
## Solution:
- Encode features ( $x_i$ ), target ( $y$ ) into states
- Compute ( $I(X_i; Y) = S(\rho_{X_i}) + S(\rho_Y) - S(\rho_{X_i Y})$ )
- Use swap test for entropies
- Select top-k features

# Practice Problem 4: QCNN

## Problem:
Design a 4-qubit quanvolutional layer for 2x2 image patch
## Solution:
- Encode 4 pixels into RY rotations
- Apply ansatz: ( $U(\theta) = R_Y(\theta_1-\theta_4) \cdot \text{CNOT}(1,2) \cdot \text{CNOT}(3,4)$ )
- Measure ( $\langle Z_i \rangle$ ) as new features
