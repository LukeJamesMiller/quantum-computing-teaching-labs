# Introduction to Quantum Graph Neural Networks
- **Definition:** Hybrid quantum-classical models that extend classical Graph Neural Networks (GNNs) using quantum circuits for graph data processing
- **Purpose:** Leverage quantum entanglement and superposition for enhanced representation and computation on graphs
- **Key Features:**
  - Decompositional approaches for scalability (e.g., ego-graphs)
  - Parameter-efficient unitaries instead of classical weights
  - Suitable for NISQ via hybrid workflows
- **Classical GNN Review:** Message passing on nodes/edges for tasks like classification
- **AI Relevance:** Processes social networks, molecules, and recommendation systems

# Graph Encoding in Quantum States
- **Concept:** Map graph nodes and edges to quantum states for processing
- **Methods:**
  - **Angle Encoding:** Node features encoded as rotation angles (e.g., RY gates), rescaled to [0, π]
  - **Bloch Sphere Mapping:** Trainable circuit maps Euclidean features to Hilbert space, preserving distances
  - **Decomposition:** Break graph into subgraphs (e.g., ego-graphs: node + neighbors) to fit limited qubits
- **Example:** For molecular graphs, encode atom features into qubit states
- **Advantage:** Exponential representation space with few qubits
- **Challenge:** Information loss in mapping; noise in state preparation

# Quantum Message Passing
- **Concept:** Quantum analog of classical message aggregation between nodes
- **Implementation:**
  - Use parameterized quantum circuits (PQC) for node/edge updates
  - Tensor products for aggregating neighbor states; unitaries (e.g., RX, RY, RZ, CNOT) for interactions
  - Iterative updates: Edge probabilities predicted via quantum layers, nodes aggregate weighted info
- **Example:** In egoQGNN, hierarchical circuits with entanglement (CNOT) and rotations for neighborhood aggregation
- **Properties:** Parameter sharing across graph; entanglement captures correlations
- **AI Use:** Learns graph representations for prediction tasks

# Applications in Social Networks or Molecular Graphs
- **Social Networks:**
  - Community detection, link prediction (e.g., friend recommendations)
  - Quantum enhancement: Entanglement models complex interactions
- **Molecular Graphs:**
  - Drug design: Classify molecules for properties (e.g., toxicity)
  - Applications: Graph classification benchmarks like MUTAG, PROTEINS
- **Other Uses:**
  - High-energy physics: Particle track reconstruction
  - Combinatorial optimization: Routing, scheduling
- **Performance:** Competitive accuracy on benchmarks, e.g., egoQGNN outperforms on PTC datasets

# Hybrid QGNN Models
- **Structure:** Quantum for subgraph processing, classical for decomposition/combination
- **Examples:**
  - DQGNN: Classical decomposition into subgraphs, quantum tensor networks for representations
  - egoQGNN: Classical control of quantum device, trainable mapping for Hilbert space
- **Benefits:**
  - Scalable to large graphs with fixed qubits
  - Classical optimization of parameters (e.g., SPSA)
- **Integration:** Quantum circuits sandwiched between classical layers
- **ML Context:** Hybrid training loops, similar to QNNs

#: Qiskit Implementations
- **Qiskit Tools:**
  - `QuantumCircuit`: Build PQCs for message passing (e.g., TwoLocal ansatz with RY, CX)
  - `Aer` Simulator: For graph classification tasks
  - Integration with Qiskit ML: Use for hybrid models, though no dedicated QGNN class
- **Workflow:**
  1. Encode graph (e.g., via NetworkX to Qiskit)
  2. Decompose into subgraphs
  3. Apply PQC per subgraph, measure expectations
  4. Classical aggregation and training
- **Examples:** Simulations in Qiskit for QSVM baselines; custom circuits for QGNN
- **Challenge:** No built-in QGNN; requires custom code
- **Lab Suggestion:** Implement simple QGNN on small graph dataset

# Scalability Analysis
- **Scalability Issues:**
  - Qubit Limit: Decompositional methods reduce to max degree D (e.g., 4-8 qubits)
  - Circuit Depth: Shallow PQCs for NISQ, but iterations increase noise
  - Parameter Count: Low (e.g., 1.68% of classical GNNs)
- **Metrics:**
  - Accuracy: Comparable to classical GNNs on benchmarks
  - Runtime: Quantum simulation slow; hardware potential for speedup
- **Mitigations:** Hybrid decomposition, noise-aware optimizers
- **Future:** Fault-tolerant for large-scale graphs
- **AI Insight:** Scalable for real-world networks (e.g., social, biological)
