# Introduction to Hybrid Methods

## Definition:
Combine quantum circuits with classical algorithms to leverage strengths of both
## Why Hybrid?
### Quantum:
Rich feature spaces, entanglement for non-linearity
### Classical:
Scalable optimization, robust training, large-scale data handling


## NISQ Suitability:
Shallow quantum circuits + classical processing reduces noise impact
## Key Paradigm:
Quantum as subroutine (e.g., feature extraction) in classical pipeline
## AI Relevance:
Enhances deep learning, optimization, and data preprocessing

---

# Quantum-Classical Interfaces

## Interfaces:
Mechanisms to pass data/results between quantum and classical systems
## Types:
### Quantum to Classical:
Measurement outcomes (e.g., expectation values, probabilities)
### Classical to Quantum:
Data encoding, parameter updates


## Challenges:
- Measurement bottleneck: Many shots needed for accuracy
- Noise in quantum outputs distorts classical processing


## Solutions:
Error mitigation (ZNE, readout correction), efficient encoding
### Example:
QNN outputs ( $\langle Z \rangle$ ) to classical loss function
## ML Context:
Interface like API between neural net layers

# Data Encoding

## Goal:
Map classical data ( x \in \mathbb{R}^d ) to quantum states
Methods:
## Amplitude Encoding:
( $|x\rangle = \sum_i x_i |i\rangle / \sqrt{\sum |x_i|^2}$ )
## Compact:
$d$-dimensional data in $\lceil \log_2 d \rceil$ qubits
## Challenge:
Complex state preparation


## Angle Encoding:
- $\prod_i R_y(x_i) |0\rangle^n$
- Simple, but one qubit per feature


## ZZ Feature Map:
Entangling rotations (from QSVM)


## Trade-Offs:
Expressivity vs. circuit depth, noise sensitivity
## AI Use:
Encode images, time-series for quantum processing

# Data Decoding

## Goal:
Extract useful information from quantum measurements
## Methods:
### Expectation Values:
( $\langle \psi(x,\theta) | Z_i | \psi(x,\theta) \rangle$ ) for classification/regression
### Probability Distributions:
Sample computational basis states
### Kernel Matrices:
Compute ( $K(x_i, x_j) = |\langle \phi(x_i) | \phi(x_j) \rangle|^2$ )


## Challenges:
### Shot noise:
Limited samples reduce precision
### Classical post-processing:
Maps quantum outputs to ML tasks


## Example:
QNN outputs ( $\langle Z_0 \rangle$ ) mapped to ${0,1}$ labels
## ML Context:
Decoding like activation function in neural nets

# Hybrid Optimization Loops

## Structure:
Quantum circuit computes loss/expectation; classical optimizer updates parameters
## Workflow:
- Encode data into quantum circuit
- Apply parameterized ansatz ( $U(\theta)$ )
- Measure to compute loss (e.g., cross-entropy, MSE)
- Update ( $\theta$ ) using classical optimizer (Adam, SPSA, COBYLA)


## Parameter Shift Rule:
Analytical gradients for quantum circuits
$\frac{\partial \langle H \rangle}{\partial \theta_i} = \frac{1}{2} [ \langle H(\theta_i + \pi/2) \rangle - \langle H(\theta_i - \pi/2) \rangle ]$


## Challenges:
Barren plateaus, noise in measurements
## AI Analogy:
Backpropagation loop in deep learning

# Theoretical Insights

## Expressivity:
Quantum circuits capture non-linear patterns via entanglement
## Complexity:
Quantum part $O(polylog(N))$, classical part dominates for large data
## Quantum Advantage:
