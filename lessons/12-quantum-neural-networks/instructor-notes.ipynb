# Introduction to Quantum Neural Networks

## Definition:
Parameterized quantum circuits acting as neural networks for ML tasks
## Purpose:
Learn complex patterns via quantum states, leveraging superposition and entanglement
## Key Features:
- Variational circuits with trainable parameters
- Hybrid quantum-classical workflow
- Suitable for NISQ devices (shallow circuits)


## Comparison to Classical NNs:
Quantum circuits as layers; weights → gate angles
## AI Relevance:
Potential for quantum advantage in non-linear data tasks

---

# Quantum Circuit Learning

## Concept:
Train quantum circuit parameters to minimize a loss function
## Components:
### Ansatz:
Parameterized circuit ( $U(\theta)$ ) producing ( $|\psi(\theta)\rangle$ )
### Loss Function:
Typically expectation value, e.g., ( $L = \langle \psi(\theta) | H | \psi(\theta) \rangle$ ), or classification loss
### Optimization:
Classical optimizer updates ( $\theta$ )


## Process:
- Encode data into quantum state
- Apply parameterized circuit
- Measure to compute loss
- Optimize parameters


## ML Analogy:
Like training weights in a neural network

---

# Variational Quantum Classifiers

## Goal:
Classify data by mapping to quantum states and measuring outputs
## Circuit Structure:
### Data encoding:
Feature map ( $U_\phi(x)$ )
### Variational ansatz:
( $U(\theta)$ ) (e.g., layers of R_y, CNOT)
### Measurement:
Expectation of Pauli operators (e.g., ( $\langle Z_0 \rangle$ ))


## Loss Function:
Binary cross-entropy or hinge loss based on measurements
### Example:
For binary classification, map ( $\langle Z \rangle$ ) to labels {0,1}
## AI Use:
Classify images, medical data, or financial patterns

---

# Variational Quantum Regressors

## Goal:
Predict continuous values using quantum circuits
## Circuit Design:
Similar to classifiers but outputs continuous expectation
### Example:
( $f(x) = \langle \psi(x,\theta) | \sum_i Z_i | \psi(x,\theta) \rangle$ )


## Loss Function:
Mean squared error between predicted and true values
## Optimization:
Minimize loss via classical optimizer
## Applications:
Time-series forecasting, regression in scientific data
## Challenge:
Precision limited by measurement shots

---

# Training with Gradient Descent

## Parameter Shift Rule:
### Compute gradients analytically:
( $\frac{\partial \langle H \rangle}{\partial \theta_i} = \frac{1}{2} [ \langle H(\theta_i + \pi/2) \rangle - \langle H(\theta_i - \pi/2) \rangle ]$ )
No finite differences needed, noise-resilient


## Optimizers:
### Gradient-based:
Adam, SGD
###  Gradient-free:
SPSA (robust to noise), COBYLA


## Challenges:
### Barren Plateaus:
Vanishing gradients for large circuits
### Noise:
Distorts expectation values


## ML Analogy:
Similar to backpropagation in classical NNs

# Comparison with Classical Neural Networks

## Classical NNs:
- Layers of neurons, activation functions (e.g., ReLU)
- Scalable, well-optimized (e.g., TensorFlow, PyTorch)
- Large datasets, deep architectures


## Quantum NNs:
- Layers of quantum gates, entanglement for non-linearity
- Limited by NISQ noise, qubit count
- Potential advantage: Exponential feature space with few qubits


## Performance:
QNNs competitive on small, complex datasets; classical better for large-scale
