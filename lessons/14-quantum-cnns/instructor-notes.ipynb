# Classical CNNs: A Quick Review

## Structure:
- Convolutional layers: Apply filters to extract features (e.g., edges)
- Pooling: Downsample to reduce dimensionality
- Fully connected layers: Classification/regression


## Key Properties:
- Local connectivity: Filters scan small regions
- Parameter sharing: Same filter across image
- Translation invariance: Robust to shifts


## Success:
Image classification, object detection (e.g., ResNet, VGG)
## Quantum Goal:
Replicate these properties with quantum circuits

# Quanvolutional Layers

## Definition:
Quantum circuit acting as a convolutional filter
## Operation:
- Input: Small patch of data (e.g., 2x2 pixels) encoded into qubits
- Circuit: Parameterized unitary ( U(\theta) ) (e.g., R_y, CNOT)
- Output: Expectation values (e.g., ( \langle Z_i \rangle )) as features


## Example (2x2 patch):
- Encode 4 pixels into 4 qubits via angle encoding
- Apply $U(\theta)$: Entangling gates + rotations
- Measure: Output 4 expectation values as new “pixels”


## Advantage:
Non-linear transformations via entanglement
## AI Use:
Feature extraction for images, sequences

---

# Parameter Sharing in Quantum Circuits

## Classical CNNs:
Same filter weights across image regions
## Quantum Analog:
- Use same $U(\theta)$ for each patch
- Parameters $\theta$ shared across circuit applications


## Implementation:
Slide quanvolutional circuit over data grid
### Example:
4x4 image → apply $U(\theta)$ to each 2x2 patch


## Benefits:
- Reduces parameter count (like classical CNNs)
- Mimics translation invariance


## Challenge:
Noise accumulates over multiple applications
## ML Context:
Efficient learning for structured data

---

# QCNN Architecture

## Structure:
- Quanvolutional Layers: Apply ( U(\theta) ) to patches
- Quantum Pooling: Reduce qubit count (e.g., measure subset, discard)
- Variational Layers: Additional parameterized circuits for learning
- Measurement: Expectation values for classification/regression


## Example (Simplified):
- Input (4x4 image)
- [Quanvolution: $U(θ)$ on 2x2 patches]
- [Pooling: Measure 1 qubit/patch]
- [Variational $U(φ)$]
- Measure ⟨Z⟩


## Depth:
$O(\text{layers * patches})$, shallow for NISQ
## AI Analogy:
Combines convolution and dense layers in quantum form

# Training QCNNs

## Process:
- Encode data patches into qubits (e.g., angle encoding)
- Apply quanvolutional and variational circuits
- Measure expectations to compute loss (e.g., cross-entropy)
- Optimize parameters ( $\theta, \phi$ ) using classical optimizer


## Optimizers:
SPSA, Adam (via parameter shift rule)
## Loss Function:
Binary cross-entropy for classification, MSE for regression
## Challenges:
- Barren plateaus in deep circuits
- Noise in measurements


## ML Connection:
Similar to CNN training with backpropagation
