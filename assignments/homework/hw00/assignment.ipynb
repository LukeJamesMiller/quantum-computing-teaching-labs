> **Dirac ↔ NumPy cheatsheet**  
> • ket `|v⟩` ↔ column vector `v` (shape `(n,1)`); bra `⟨v|` ↔ conjugate transpose `v.conj().T`  
> • inner product `⟨u|v⟩ = u† v` (complex scalar) • projector `|u⟩⟨u|` (matrix)  
> • expectation `⟨A⟩ = ⟨ψ|A|ψ⟩` (pure) or `Tr(Aρ)` (mixed)
